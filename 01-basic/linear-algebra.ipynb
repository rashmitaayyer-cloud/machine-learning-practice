### Vector Vector Multiplication

```python
u = np.array([2, 4, 5, 6], dtype=np.float64)
```

```python
v = np.array([1, 2, 3, 4], dtype=np.float64)
```

```python
# In Python
def vector_vector_multiplication(u, v):
    assert u.shape[0] == v.shape[0]

    n = u.shape[0]
    result = 0.0
        
    for i in range(n):
        result = result + (u[i] * v[i])

    return result
```

```python
vector_vector_multiplication(u, v)
```

```python
# In numpy
np.dot(u, v)
```

### Matrix Vector Multiplication

```python
A = np.arange(12, dtype=np.float64).reshape(3,4)
```

```python
A
```

```python
b = np.arange(4)
```

```python
# In python
def matrix_vector_multiplication(A, b):
    assert A.shape[1] == b.shape[0]

    num_rows = A.shape[0]
    result = np.zeros(num_rows)

    for i in range(num_rows):
        result[i] = vector_vector_multiplication(A[i], b)

    return result
```

```python
matrix_vector_multiplication(A, b)
```

```python
# In numpy
np.dot(A,b)
```

```python
V = np.arange(12).reshape(4, 3)
V.shape[0]
```

```python
def matrix_matrix_multiplication(X, Y):
    assert X.shape[1] == Y.shape[0]

    num_rows = X.shape[0]
    num_cols = Y.shape[1]

    result = np.zeros((num_rows, num_cols))

    for i in range(num_cols):
        yi = Y[:, i]
        Xyi = matrix_vector_multiplication(X, yi)
        result[:,i] = Xyi

    return result
    
```

```python
matrix_matrix_multiplication(A, V)
```

```python
np.dot(A, V)
```
