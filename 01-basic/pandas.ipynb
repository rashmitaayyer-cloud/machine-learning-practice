```python
import pandas as pd
import numpy as np
```

##### Creating a Series

```python
s = pd.Series([1, -3, 9.5])
s
```

```python
# Series behave like numpy arrays, so we can do elementwise operations on them
np.exp(s)
```

```python
# Broadcasting
s + 4000
```

##### Index Labels

```python
s2 = pd.Series([2, 3, 32, -8, -5.4], index=['a', 'b', 'c', 'd', 'e'])
s2
```

```python
# 'loc' attribute when accessing labels and 'iloc' attribute when accessing by integer allocation
s2.loc["c"]
s2.iloc[0]
```

```python
s2.iloc[1:3]
```

```python
new = pd.Series([1000, 2000, 3000, 4000])
new
```

```python
new_sliced = new[2:]
new_sliced
```

```python
try:
    new_sliced[0]
except KeyError as e:
    print("KeyError", e)
```

```python
# But we can access it with iloc attribute
new_sliced.iloc[0]
```

### Init from `dict`

We can create `Series` from `dict`. The `keys` will be used as index labels:

```python
weights = {"alice" : 56, "bob" : 63, "lyca" : 63}
s3 = pd.Series(weights)
s3
```

```python
""" You can control which elements you want to include in the `Series`
and in what order by explicitly specifying the desired `index`:"""
s4 = pd.Series(weights, index = ['lyca', 'bob'])
s4
```

### Automatic Alignment

```python
print(s2.keys())
print(s3.keys())

s2 + s3
```
